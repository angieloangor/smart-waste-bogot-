# Limpieza y Normalización Espacial
Convertimos los CRS a EPSG:3116 y limpiamos valores nulos/incorrectos.

```python
import geopandas as gpd
import pandas as pd
import os

RAW_DIR = '../01_DATOS/01_RAW'
PROC_DIR = '../01_DATOS/02_PROCESADOS'

# Cargar
puntos = gpd.read_file(f"{RAW_DIR}/puntos_criticos_arrojo_clandestino_residuos.geojson")
cestas = gpd.read_file(f"{RAW_DIR}/cestas.gpkg")
macrorutas = gpd.read_file(f"{RAW_DIR}/macrorutas_de_barrido.gpkg")
reportes = gpd.read_file(f"{RAW_DIR}/Geopackage.gpkg")
```

```python
# Normalizar CRS a EPSG:3116
TARGET_CRS = 'EPSG:3116'
puntos_3116 = puntos.to_crs(TARGET_CRS)
cestas_3116 = cestas.to_crs(TARGET_CRS)
macrorutas_3116 = macrorutas.to_crs(TARGET_CRS)
reportes_3116 = reportes.to_crs(TARGET_CRS)
```

```python
# Limpieza básica
# Eliminar nulos en geometría
reportes_3116 = reportes_3116.dropna(subset=['geometry'])
reportes_3116 = reportes_3116[reportes_3116.is_valid]

puntos_3116 = puntos_3116.dropna(subset=['geometry'])
puntos_3116 = puntos_3116[puntos_3116.is_valid]

cestas_3116 = cestas_3116.dropna(subset=['geometry'])
cestas_3116 = cestas_3116[cestas_3116.is_valid]
```

```python
# Guardar limpios
puntos_3116.to_file(f"{PROC_DIR}/puntos_limpios.gpkg", driver='GPKG')
cestas_3116.to_file(f"{PROC_DIR}/cestas_limpias.gpkg", driver='GPKG')
reportes_3116.to_file(f"{PROC_DIR}/reportes_limpios.gpkg", driver='GPKG')
macrorutas_3116.to_file(f"{PROC_DIR}/macrorutas_limpias.gpkg", driver='GPKG')

print("Datos limpios y normalizados exportados a 02_PROCESADOS.")
```

# Exploración de Datos: Bogotá Residuos Inteligente
En este notebook realizaremos una exploración inicial de los datasets base para entender su estructura, valores nulos y distribuciones.

```python
import geopandas as gpd
import pandas as pd
import matplotlib.pyplot as plt
import os

# Definir rutas
RAW_DIR = '../01_DATOS/01_RAW'
```

```python
# Cargar datos
puntos_criticos = gpd.read_file(f"{RAW_DIR}/puntos_criticos_arrojo_clandestino_residuos.geojson")
cestas = gpd.read_file(f"{RAW_DIR}/cestas.gpkg")
macrorutas = gpd.read_file(f"{RAW_DIR}/macrorutas_de_barrido.gpkg")
reportes = gpd.read_file(f"{RAW_DIR}/Geopackage.gpkg")

print("Puntos críticos:", puntos_criticos.shape)
print("Cestas:", cestas.shape)
print("Macrorutas:", macrorutas.shape)
print("Reportes:", reportes.shape)
```

```python
# Información de Reportes
reportes.info()
print("\nNulos en reportes:\n", reportes.isna().sum())
print("\nDuplicados en reportes:", reportes.duplicated().sum())
```
